```python

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values percentage:")
missing_percent = (df.isnull().sum() / len(df)) * 100
print(missing_percent.round(2))

print("\nData types:")
print(df.dtypes)
```

```python

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Total rows before cleaning:", len(df))

# Display important IPL columns if available
important_columns = [
    'match_id',
    'match_winner',
    'toss_winner',
    'venue',
    'player_of_match',
    'winner'
]

for col in important_columns:
    if col in df.columns:
        print(f"\n{col}: Available")
    else:
        print(f"\n{col}: Not available")

print("\nUpdated column names:")
print(df.columns.tolist())
```

```python

# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Extract year
df['year'] = df['date'].dt.year

# Count matches per year
matches_per_year = df['year'].value_counts().sort_index()

print("Number of Matches Played Each Year:")
display(matches_per_year)

# Visualize
plt.figure(figsize=(10, 5))
sns.barplot(x=matches_per_year.index, y=matches_per_year.values, color='teal')

plt.title("Number of IPL Matches Played Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Count matches won by each team
match_wins = df['match_winner'].dropna().value_counts()

print("Matches Won by Each Team:")
display(match_wins)

# Create bar chart
plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=match_wins.values,
    y=match_wins.index,
    color='teal'
)

plt.title("Most Matches Won by Each IPL Team", fontsize=14)
plt.xlabel("Number of Matches Won")
plt.ylabel("Team")

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Count toss decisions
toss_decisions = df['toss_decision'].dropna().value_counts()

print("Toss Decision Counts:")
display(toss_decisions)

# Calculate percentages
toss_percentage = (
    df['toss_decision'].dropna().value_counts(normalize=True) * 100
).round(2)

print("\nToss Decision Percentage:")
display(toss_percentage)

# Plot
plt.figure(figsize=(7, 5))

ax = sns.barplot(
    x=toss_decisions.index,
    y=toss_decisions.values,
    color='orange'
)

plt.title("IPL Toss Decisions: Batting vs Fielding")
plt.xlabel("Toss Decision")
plt.ylabel("Number of Matches")

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Keep rows where toss winner and match winner are available
toss_match_df = df.dropna(subset=['toss_winner', 'match_winner']).copy()

# Compare toss winner with match winner
toss_match_df['toss_and_match_winner_same'] = (
    toss_match_df['toss_winner'] == toss_match_df['match_winner']
)

toss_wins_match = toss_match_df['toss_and_match_winner_same'].sum()
total_valid_matches = len(toss_match_df)

toss_win_percentage = (
    toss_wins_match / total_valid_matches * 100
    if total_valid_matches > 0 else 0
)

print("Matches where toss winner also won the match:", toss_wins_match)
print("Total matches with both winners available:", total_valid_matches)
print(f"Toss winner also won the match: {toss_win_percentage:.2f}%")

print(f"Toss winner did not win the match: {100 - toss_win_percentage:.2f}%")
```

```python

player_awards = df['player_of_the_match'].dropna().value_counts()

print("Top 10 Players by Player of the Match Awards:")
display(player_awards.head(10))

# Visualize top 10 players
top_10_players = player_awards.head(10)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_10_players.values,
    y=top_10_players.index,
    color='teal'
)

plt.title("Top 10 Players by Player of the Match Awards")
plt.xlabel("Number of Awards")
plt.ylabel("Player")

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Count how many times each player was listed as top scorer
top_scorers = df['top_scorer'].dropna().value_counts()

print("Top 10 Batsmen by Top-Scorer Appearances:")
display(top_scorers.head(10))

# Visualize the top 10
top_10_scorers = top_scorers.head(10)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_10_scorers.values,
    y=top_10_scorers.index,
    color='orange'
)

plt.title("Top 10 Batsmen by Top-Scorer Appearances")
plt.xlabel("Number of Appearances")
plt.ylabel("Batsman")

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Count wins by winning method
win_methods = df['won_by'].dropna().value_counts()

print("Matches Won by Method:")
display(win_methods)

# Plot the results
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=win_methods.index,
    y=win_methods.values,
    color='orange'
)

plt.title("IPL Matches Won by Runs vs Wickets")
plt.xlabel("Winning Method")
plt.ylabel("Number of Matches")
plt.xticks(rotation=15)

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Count best bowling performance appearances
best_bowlers = df['best_bowling'].dropna().value_counts()

print("Top 10 Best Bowling Performers:")
display(best_bowlers.head(10))

# Plot top 10
top_10_bowlers = best_bowlers.head(10)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_10_bowlers.values,
    y=top_10_bowlers.index,
    color='teal'
)

plt.title("Top 10 Best Bowling Performance Appearances")
plt.xlabel("Number of Appearances")
plt.ylabel("Bowler")

ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python

# Convert innings scores to numeric values
df['first_ings_score'] = pd.to_numeric(
    df['first_ings_score'], errors='coerce'
)

df['second_ings_score'] = pd.to_numeric(
    df['second_ings_score'], errors='coerce'
)

# Calculate average scores
first_avg = df['first_ings_score'].mean()
second_avg = df['second_ings_score'].mean()

print(f"Average First Innings Score: {first_avg:.2f}")
print(f"Average Second Innings Score: {second_avg:.2f}")

# Compare the averages
score_comparison = pd.DataFrame({
    'Innings': ['First Innings', 'Second Innings'],
    'Average Score': [first_avg, second_avg]
})

plt.figure(figsize=(7, 5))

ax = sns.barplot(
    data=score_comparison,
    x='Innings',
    y='Average Score',
    color='teal'
)

plt.title("Average Score: First Innings vs Second Innings")
plt.xlabel("Innings")
plt.ylabel("Average Runs")

ax.bar_label(ax.containers[0], fmt='%.1f', padding=3)

plt.tight_layout()
plt.show()
```

```python

# Convert margin to numeric
df['margin_numeric'] = pd.to_numeric(df['margin'], errors='coerce')

# Separate matches won by runs and wickets
runs_wins = df.loc[
    df['won_by'].astype(str).str.contains('run', case=False, na=False),
    'margin_numeric'
].dropna()

wicket_wins = df.loc[
    df['won_by'].astype(str).str.contains('wicket', case=False, na=False),
    'margin_numeric'
].dropna()

print("Average winning margin (runs):", round(runs_wins.mean(), 2))
print("Average winning margin (wickets):", round(wicket_wins.mean(), 2))

# Plot distributions separately
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(runs_wins, bins=10, kde=True, ax=axes[0])
axes[0].set_title("Winning Margin in Runs")
axes[0].set_xlabel("Runs")

sns.histplot(wicket_wins, bins=10, kde=True, ax=axes[1])
axes[1].set_title("Winning Margin in Wickets")
axes[1].set_xlabel("Wickets")

plt.tight_layout()
plt.show()
```

```python
team_scores = pd.concat([
    df[['team1', 'first_ings_score']].rename(
        columns={'team1': 'team', 'first_ings_score': 'score'}
    ),
    df[['team2', 'second_ings_score']].rename(
        columns={'team2': 'team', 'second_ings_score': 'score'}
    )
], ignore_index=True)

team_scores['score'] = pd.to_numeric(
    team_scores['score'], errors='coerce'
)

average_scores = (
    team_scores.dropna(subset=['team', 'score'])
    .groupby('team')['score']
    .mean()
    .sort_values(ascending=False)
)

print("Team-wise Average Score:")
print(average_scores.round(2))

plt.figure(figsize=(12, 6))
sns.barplot(x=average_scores.values, y=average_scores.index)
plt.title("Team-wise Average Innings Score")
plt.xlabel("Average Score")
plt.ylabel("Team")
plt.tight_layout()
plt.show()
```

```python
team_toss_decision = (
    df.dropna(subset=['toss_winner', 'toss_decision'])
    .groupby(['toss_winner', 'toss_decision'])
    .size()
    .unstack(fill_value=0)
)

print("Team-wise Toss Decisions:")
print(team_toss_decision)

team_toss_decision.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title("Team-wise Toss Decisions")
plt.xlabel("Team")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45, ha='right')
plt.legend(title="Toss Decision")
plt.tight_layout()
plt.show()
```

```python
df['month'] = df['date'].dt.month

monthly_matches = df['month'].value_counts().sort_index()

month_names = {
    1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr',
    5: 'May', 6: 'Jun', 7: 'Jul', 8: 'Aug',
    9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'
}

monthly_matches.index = monthly_matches.index.map(month_names)

print("Matches played per month:")
print(monthly_matches)

plt.figure(figsize=(10, 5))
sns.barplot(x=monthly_matches.index, y=monthly_matches.values)
plt.title("Monthly Distribution of Matches")
plt.xlabel("Month")
plt.ylabel("Number of Matches")
plt.tight_layout()
plt.show()
```

```python
year_team_wins = (
    df.dropna(subset=['year', 'match_winner'])
    .groupby(['year', 'match_winner'])
    .size()
    .reset_index(name='wins')
)

year_team_wins['year'] = year_team_wins['year'].astype(int)

top_team_each_year = year_team_wins.loc[
    year_team_wins.groupby('year')['wins'].idxmax()
].sort_values('year')

print("Top-performing team in each year:")
print(top_team_each_year.to_string(index=False))

year_team_pivot = year_team_wins.pivot(
    index='year',
    columns='match_winner',
    values='wins'
).fillna(0)

year_team_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7),
    legend=False
)

plt.title("Team-wise Match Wins Across Years")
plt.xlabel("Year")
plt.ylabel("Number of Wins")
plt.tight_layout()
plt.show()
```

```python
# Total matches played by each team
team_matches = pd.concat([
    df[['team1']].rename(columns={'team1': 'team'}),
    df[['team2']].rename(columns={'team2': 'team'})
])

matches_played = team_matches['team'].value_counts()

# Total matches won by each team
matches_won = df['match_winner'].value_counts()

# Combine the results
team_performance = pd.DataFrame({
    'Matches Played': matches_played,
    'Matches Won': matches_won
}).fillna(0)

team_performance['Win Percentage'] = (
    team_performance['Matches Won'] /
    team_performance['Matches Played'] * 100
).round(2)

team_performance = team_performance.sort_values(
    'Win Percentage', ascending=False
)

print("Team Performance:")
print(team_performance)

plt.figure(figsize=(12, 6))
sns.barplot(
    x=team_performance['Win Percentage'],
    y=team_performance.index
)
plt.title("Team-wise Win Percentage")
plt.xlabel("Win Percentage (%)")
plt.ylabel("Team")
plt.tight_layout()
plt.show()
```

```python
# Convert scores to numeric values
score_columns = ['first_ings_score', 'second_ings_score']

for col in score_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Keep rows where both scores are available
score_data = df[score_columns].dropna()

# Calculate correlation
correlation = score_data['first_ings_score'].corr(
    score_data['second_ings_score']
)

print("Correlation between first and second innings scores:",
      round(correlation, 3))

# Scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=score_data,
    x='first_ings_score',
    y='second_ings_score'
)
plt.title("First Innings Score vs Second Innings Score")
plt.xlabel("First Innings Score")
plt.ylabel("Second Innings Score")
plt.tight_layout()
plt.show()
```

```python
# Select matches with the highest first innings scores
top_scoring_matches = df.nlargest(
    10, 'first_ings_score'
)[[
    'match_id',
    'date',
    'venue',
    'team1',
    'team2',
    'first_ings_score',
    'match_winner'
]]

print("Top 10 Highest-Scoring Matches:")
print(top_scoring_matches.to_string(index=False))

# Visualize the scores
plot_data = top_scoring_matches.sort_values('first_ings_score')

plt.figure(figsize=(12, 7))
sns.barplot(
    data=plot_data,
    x='first_ings_score',
    y='match_id'
)
plt.title("Top 10 Matches by First Innings Score")
plt.xlabel("First Innings Score")
plt.ylabel("Match ID")
plt.tight_layout()
plt.show()
```

```python
venue_team_wins = (
    df.dropna(subset=['venue', 'match_winner'])
    .groupby(['venue', 'match_winner'])
    .size()
    .reset_index(name='wins')
)

# Find the most wins by a team at each venue
top_venue_teams = venue_team_wins.loc[
    venue_team_wins.groupby('venue')['wins'].idxmax()
].sort_values('wins', ascending=False)

print("Top team at each venue:")
print(top_venue_teams.to_string(index=False))

# Show top 10 venue-team combinations
top_10_venue_teams = top_venue_teams.head(10).copy()
top_10_venue_teams['venue_team'] = (
    top_10_venue_teams['venue'] + " — " +
    top_10_venue_teams['match_winner']
)

plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_10_venue_teams,
    x='wins',
    y='venue_team'
)
plt.title("Most Wins by a Team at Each Venue")
plt.xlabel("Number of Wins")
plt.ylabel("Venue — Team")
plt.tight_layout()
plt.show()
```

```python
# Keep matches with valid toss and match winners
decision_analysis = df.dropna(
    subset=['toss_winner', 'toss_decision', 'match_winner']
).copy()

# Did the toss-winning team also win the match?
decision_analysis['toss_winner_won_match'] = (
    decision_analysis['toss_winner'] ==
    decision_analysis['match_winner']
)

# Calculate wins and match counts for each toss decision
decision_summary = decision_analysis.groupby(
    'toss_decision'
).agg(
    Total_Matches=('match_id', 'count'),
    Toss_Winner_Match_Wins=('toss_winner_won_match', 'sum')
)

decision_summary['Win_Percentage'] = (
    decision_summary['Toss_Winner_Match_Wins'] /
    decision_summary['Total_Matches'] * 100
).round(2)

print("Toss Decision vs Match Result:")
print(decision_summary)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=decision_summary.reset_index(),
    x='toss_decision',
    y='Win_Percentage'
)
plt.title("Toss Winner's Match Win Percentage by Decision")
plt.xlabel("Toss Decision")
plt.ylabel("Win Percentage (%)")
plt.tight_layout()
plt.show()
```

```python
player_award_analysis = (
    df.dropna(subset=['player_of_the_match', 'match_winner'])
    .groupby(['player_of_the_match', 'match_winner'])
    .size()
    .reset_index(name='count')
    .sort_values('count', ascending=False)
)

print("Player of the Match and Match Winner:")
print(player_award_analysis.head(15).to_string(index=False))

top_players = (
    df['player_of_the_match']
    .dropna()
    .value_counts()
    .head(10)
    .index
)

top_player_winners = player_award_analysis[
    player_award_analysis['player_of_the_match'].isin(top_players)
]

plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_player_winners,
    x='count',
    y='player_of_the_match',
    hue='match_winner'
)
plt.title("Player of the Match Awards by Match Winner")
plt.xlabel("Number of Matches")
plt.ylabel("Player")
plt.legend(title="Match Winner", bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

This IPL Match Data Analysis project explores match results, team performance, toss decisions, player achievements, venues, innings scores, and winning margins using Python.

The dataset contains 74 matches played between March 26, 2022, and May 29, 2022. Data cleaning was performed to handle date formatting issues, and the final quality check confirmed zero missing dates and zero duplicate rows.

This project demonstrates practical skills in Python, Pandas, NumPy, Matplotlib, and Seaborn for data cleaning, exploratory data analysis, statistical summaries, and visualization.

The findings are limited to the matches included in the dataset and should not be interpreted as representing the entire history of the IPL.
